```python
import pandas as pd

df = pd.read_csv('./ableton_synth_preset_list.csv')
```

```python
df.head()
```

```python
d_out = {}
for i in range(len(df)):
    id = df['id'].loc[i]
    name = df['name'].loc[i]
    d_out[id] = name
```

```python
d_out
import json

with open('./id_name_lookup.json', 'w') as f:
    json.dump(d_out, f)
```
